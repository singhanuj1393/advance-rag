```python
from langchain_community.document_loaders.web_base import WebBaseLoader
from pprint import pp
```

```python
# URLs to load — WebBaseLoader accepts multiple URLs at once
url_1 = "https://reference.langchain.com/python/langchain-community/document_loaders/web_base/WebBaseLoader"
url_2 = "https://reference.langchain.com/python/langchain-community/document_loaders/pdf/PyPDFLoader"
url_3 = "https://reference.langchain.com/python/langchain-community/document_loaders/csv_loader/CSVLoader"

# Tip: for a single URL use web_path="<url>" instead of web_paths=[...]
loader = WebBaseLoader(
    web_paths=[url_1, url_2, url_3],  # list of URLs to scrape
    show_progress=True,               # shows a progress bar while loading
)
```

```python
# load all URLs — returns one Document per URL
document = loader.load()
```

```python
# each URL becomes one Document, so total documents == number of URLs passed
print(len(document))
```

```python
# metadata of the first document (source URL, title, description, etc.)
pp(f"metadata: {document[0].metadata}")
```

```python
# raw page content of the first document
print(document[0].page_content)
```

```python
# metadata of the second document
pp(document[1].metadata)
```

## Recursive URL Loader
#### Crawls a root URL and follows all child links up to a given depth.
#### Useful when you want to load an entire docs site, not just a single page.
* for more : [Recursive URL Loader](https://reference.langchain.com/python/langchain-community/document_loaders/recursive_url_loader/RecursiveUrlLoader)

```python
from langchain_community.document_loaders import RecursiveUrlLoader
from pprint import pp
```

```python
# root URL to start crawling from
base_url = "https://reference.langchain.com/python/langchain-community/document_loaders"

recursive_loader = RecursiveUrlLoader(
    url=base_url,
    max_depth=2  # how many levels of child links to follow (1 = root only, 2 = root + its direct children)
)
```

```python
# lazy_load() returns a generator instead of loading everything into memory at once
# use this when the site is large — documents are fetched one by one on demand
# documents = recursive_loader.load()  # <-- avoid for large sites, loads all at once
lazy_documents = recursive_loader.lazy_load()
```

```python
lazy_documents  # generator object — nothing is fetched yet until you iterate
```

```python
# iterate through the generator and print the first 15 documents
# stops early so we don't wait for the entire site to load
counter = 0
for doc in lazy_documents:
    if counter == 15:
        break
    counter += 1
    print(f"metadata : {doc.metadata}")
    print(f"{doc.page_content[:300]}\n")
```
