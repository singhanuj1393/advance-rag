```python
from langchain_community.document_loaders import CSVLoader
from pprint import pp
from pathlib import Path
```

```python
## define csv file path
file_path = Path("../knowledge-source/organizations.csv")
```

```python
## check if path is correct and file exists

print(file_path.exists())
```

```python
## create csv loader 

loader = CSVLoader(file_path=file_path.as_posix(),
                   metadata_columns=['Organization Id','Founded','Number of employees'],
                   source_column='Industry',
                   content_columns=['Description']
                   )
```

```python
## load the file

csv_documents = loader.load()
```

```python
## print documents
## Note: each row consider as one document in csv
print(f"document length: ",len(csv_documents))
print(f"metadata info : ",csv_documents[0].metadata)
print(f"page content info : \n \n",csv_documents[0].page_content)
```

###  Important NOTE
`metadata_columns` — Stores these CSV columns as document metadata.

`Organization Id`
`Founded`
`Number of employees`

`source_column` — Uses `Industry` as the document source.

`content_columns` — Uses `Description` as the main document content.
 * Defines which column values should be embedded for similarity search. Use it to embed only relevant values and improve search accuracy.

If `content_columns` is not defined: All remaining CSV column values are included in the document content, except the columns listed in `metadata_columns`.
